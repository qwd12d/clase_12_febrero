COMPLEJIDAD ALGORITMICA: 
Cantidad de recurdos (temporales) que necesita un algoritmo para resolver un problema y por tanto permite determinar la eifciente de dicho algoritmo

Medida independiente 
- Maquina
- Lenguaje de programación
- Compilador
- Elemento de hardware o sorfware

Solución: 
Determinar cuantos pasos se efectuan al ejecutarse el algoritmo

Consideraciones:
- El costo depende del tamaño de los datos
- Peor caso (Importante)
- Caso promedio
- Mejor caso

Tamaño de los datos
- Número en si
- Número de digitos
- Cantidad de elementos

Big O Notacion - Peor caso
- Se usa para caracterizar la complejidad

```python
## Ejemplo de Big O Notation en Python
def obtener_primer_elemento(lista):
    return lista[0]

def es_par(numero):
    return numero % 2 == 0

def valor_absoluto(numero):
    if numero < 0:
        return -numero
    return numero

def encontrar_maximo(lista):            # (O(n)) = 1+n*(1+1)+1
    maximo = lista[0] #1                         = 2+2n (se cancelan los terminos constantes)
    for numero in lista: #n                      = n
        if numero > maximo: #1
            maximo = numero #1
    return maximo #1

def contar_apariciones(lista, numero_buscado):
    contador = 0 #1
    i = 0 #1                                    # (O(n)) = 1+1+n*(1+1+1)+1
    while i < len(lista): #n                             = 3+3n (se cancelan los terminos constantes)
        if lista[i] == numero_buscado: #1                = n
            contador += 1 #1
        i += 1 #1
    return contador #1

def potencia(base, exponente):              # 
    resultado = 1 #1
    while exponente > 0: #                  exp/2^k = 1 (despejar)
        if exponente % 2 == 1: #1           exp = 2^k
            resultado *= base #1            k = log2(exp)
        base *= base #1
        exponente //= 2 #1
    return resultado #1
```
