Ejericio 4:

```python
def llego_tarde_elegante(llegadas, nombre):
    posicion = llegadas.index(nombre)
    return posicion >= len(llegadas) / 2 and posicion != len(llegadas) - 1

fiesta = ['Adela', 'Fleda', 'Owen', 'May', 'Mona', 'Gilbert', 'Ford']

for invitado in fiesta:
    resultado = "SI" if llego_tarde_elegante(fiesta, invitado) else "NO"
    print(invitado +" : " + resultado)

print()
nombre = input("Verificar invitado: ")
resultado = "SI, llego tarde elegante" if llego_tarde_elegante(fiesta, nombre) else "NO llego tarde elegante"
print(resultado)
```

```python
def fizzbuzz(inicio, fin, div1, div2, pal1, pal2):
    for n in range(inicio, fin + 1):
        if n % div1 == 0 and n % div2 == 0:
            print(f"{n}: {pal1}{pal2}")
        elif n % div1 == 0:
            print(f"{n}: {pal1}")
        elif n % div2 == 0:
            print(f"{n}: {pal2}")
        else:
            print(n)

inicio = int(input("Número inicial: "))
fin = int(input("Número final: "))
div1 = int(input("Primer divisor: "))
div2 = int(input("Segundo divisor: "))
pal1 = input("Primera palabra: ")
pal2 = input("Segunda palabra: ")

print()
fizzbuzz(inicio, fin, div1, div2, pal1, pal2)
```

Ejercicio 7:

```python
numero = int(input("Ingrese un numero: "))

def suma_digitos(n):
    suma = 0
    while n > 0:
        suma += n % 10
        n //= 10
    return suma

resultado = suma_digitos(numero)
print(f"La suma de los digitos de {numero} es {resultado}")
```

Ejericicio 10:

```python
def contar_vocales_consonantes(texto):
    vocales = "aeiouAEIOU"
    num_vocales = 0
    num_consonantes = 0
    
    for caracter in texto:
        if caracter.isalpha():
            if caracter in vocales:
                num_vocales += 1
            else:
                num_consonantes += 1
    
    return num_vocales, num_consonantes
texto = input("ingrese el texto: ")
vocales, consonantes = contar_vocales_consonantes(texto)
print("Texto:", texto)
print("Vocales:", vocales)
print("Consonantes:", consonantes)
```

Ejercicio 11:

```python
def encontrar_duplicados(lista):
    duplicados = []
    
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            if lista[i] == lista[j]:
                if lista[i] not in duplicados:
                    duplicados.append(lista[i])
    
    return duplicados

cantidad = int(input("¿Cuántos números vas a ingresar? "))

numeros = []

for i in range(cantidad):
    num = int(input("Ingresa un número: "))
    numeros.append(num)

resultado = encontrar_duplicados(numeros)

if len(resultado) > 0:
    print("Los números duplicados son:", resultado)
else:
    print("No hay números duplicados.")
```

Ejercicio 13:

```python
def elementos_comunes(lista1, lista2):
    comunes = []
    
    for i in range(len(lista1)):
        if lista1[i] in lista2:
            if lista1[i] not in comunes:
                comunes.append(lista1[i])
    
    return comunes

cantidad1 = int(input("Cuantos numeros tiene la primera lista: "))
lista1 = []

for i in range(cantidad1):
    num = int(input("Digite un numero: "))
    lista1.append(num)


cantidad2 = int(input("Cuantos numeros tiene la segunda lista: "))
lista2 = []

for i in range(cantidad2):
    num = int(input("Digite un numero: "))
    lista2.append(num)


resultado = elementos_comunes(lista1, lista2)

if len(resultado) > 0:
    print("Numeros que se repiten en las dos listas:", resultado)
else:
    print("No hay numeros iguales.")
```

Ejercicio 15:

```python
def contar_palabras(texto):
    conteo = {}
    texto = texto.lower()
    
    palabra = ""
    
    for letra in texto:
        if letra != " ":
            palabra = palabra + letra
        else:
            if palabra != "":
                if palabra in conteo:
                    conteo[palabra] = conteo[palabra] + 1
                else:
                    conteo[palabra] = 1
                palabra = ""
    
    if palabra != "":
        if palabra in conteo:
            conteo[palabra] = conteo[palabra] + 1
        else:
            conteo[palabra] = 1
    
    return conteo

texto_usuario = input("Escriba un texto: ")

resultado = contar_palabras(texto_usuario)

print("Cantidad de veces que aparece cada palabra:")
print(resultado)
```

Ejercicio 16:

```python
def son_anagramas(palabra1, palabra2):
    palabra1 = palabra1.lower().replace(" ", "")
    palabra2 = palabra2.lower().replace(" ", "")
    
    if len(palabra1) != len(palabra2):
        return False
    
    for letra in palabra1:
        if palabra1.count(letra) != palabra2.count(letra):
            return False
    
    return True

palabra1 = input("Ingresa la primera palabra: ")
palabra2 = input("Ingresa la segunda palabra: ")

if son_anagramas(palabra1, palabra2):
    print("SI son anagramas")
else:
    print("NO son anagramas")
```
